```python
import mglearn
from sklearn.model_selection import train_test_split

from sklearn.linear_model import LinearRegression
from sklearn.linear_model import Ridge
from sklearn.linear_model import Lasso

import numpy as np
import matplotlib.pyplot as plt
```

```python
X, y = mglearn.datasets.load_extended_boston()
```

```python
x_train, x_test, y_train, y_test = train_test_split(
    X, 
    y, 
    random_state=0)
```

```python
lr = LinearRegression().fit(x_train, y_train)
lr
```

```python
#overfitting

#we need to control complexity
print(lr.score(x_train,y_train))
print(lr.score(x_test,y_test))
```

```python
#L2 regularization
```

```python
ridge = Ridge().fit(x_train,y_train)
ridge
```

```python
print(ridge.score(x_train,y_train))
print(ridge.score(x_test,y_test))
```

```python
ridge10 = Ridge(alpha=10).fit(x_train,y_train)
ridge10
```

```python
print(ridge10.score(x_train,y_train))
print(ridge10.score(x_test,y_test))
```

```python
ridge01 = Ridge(alpha=0.1).fit(x_train,y_train)
ridge01
```

```python
print(ridge01.score(x_train,y_train))
print(ridge01.score(x_test,y_test))
```

```python
print(ridge.score(x_train,y_train))
print(ridge.score(x_test,y_test))
```

```python
plt.plot(ridge.coef_, 's', label='ridge alpha=1')
plt.plot(ridge10.coef_, '^',label='ridge alpha=10')
plt.plot(ridge01.coef_, 'v',label='ridge alpha=0.1')
plt.plot(lr.coef_, 'o',label='linear regression')

plt.hlines(0,0,len(lr.coef_))

plt.legend()
plt.ylim(-25,25)

print(len(lr.coef_))
```

```python
mglearn.plots.plot_ridge_n_samples()
```

```python
#ridge es mejor con menos data pero no lo suficiente
#more data is better!
```

```python
lasso = Lasso().fit(x_train,y_train)
lasso
```

```python
#underfitting
print(lasso.score(x_train,y_train))
print(lasso.score(x_test,y_test))
print(np.sum(lasso.coef_ != 0))
```

```python
#si decreses alpha el modelo se hace mas complejo si 
#haces que los parametros crezcan
lasso001 = Lasso(alpha=0.01, max_iter=100000).fit(x_train,y_train)
lasso001
```

```python
print(lasso001.score(x_train,y_train))
print(lasso001.score(x_test,y_test))
print(np.sum(lasso001.coef_ != 0))
```

```python
print(ridge01.score(x_train,y_train))
print(ridge01.score(x_test,y_test))
```

```python
lasso00001 = Lasso(alpha=0.0001, max_iter=100000).fit(x_train,y_train)
lasso00001
```

```python
#overfitting
print(lasso00001.score(x_train,y_train))
print(lasso00001.score(x_test,y_test))
print(np.sum(lasso00001.coef_ != 0))
```
